```python
# Find the average ride distance for each vehicle type:
df[df['Booking_Status'] == 'Success'].groupby('Vehicle_Type')['Ride_Distance'].mean().round(2).sort_values(ascending=False)
```

```python
# Get the total number of cancelled rides by customers:
(df['Booking_Status'] == 'Canceled by Driver').sum()
```

```python
# List the top 5 customers who booked the highest number of rides:
df['Customer_ID'].value_counts().head(5)
```

```python
df['Canceled_Rides_by_Driver'].unique()
```

```python
# Get the number of rides cancelled by drivers due to personal and car-related issues:
len(df[df['Canceled_Rides_by_Driver'] == 'Personal & Car related issue'])
```

```python
(df['Canceled_Rides_by_Driver'].value_counts(normalize=True) * 100).round(2)
```

```python
df['Hour'] = pd.to_datetime(df['Time'], format='%H:%M:%S').dt.hour

(df.groupby('Hour')['Canceled_Rides_by_Driver']
   .value_counts(normalize=True) * 100).round(2)
```

```python
bins = [0, 4, 6, 12, 17, 21, 24]
labels = ['Midnight', 'Early Morning', 'Morning', 'Afternoon', 'Evening', 'Night']

df['Time_Slot'] = pd.cut(df['Hour'], bins=bins, labels=labels, right=False)
```

```python
(df.groupby(['Time_Slot', 'Vehicle_Type'], observed=True)['Canceled_Rides_by_Driver']
   .value_counts(normalize=True) * 100).round(2).unstack(fill_value=0)
```

```python
# Find the maximum and minimum driver ratings for Prime Sedan bookings:
df[df['Vehicle_Type'] == 'Prime Sedan']['Driver_Ratings'].agg(['max', 'min'])
```

```python
# Retrieve all rides where payment was made using UPI:
len(df[df['Payment_Method'] == 'UPI'])
```

```python
(df['Payment_Method'].value_counts(normalize=True) * 100).round(2)
```

```python
# Find the average customer rating per vehicle type:
df.groupby('Vehicle_Type')['Customer_Rating'].mean().round(2)
```

```python
# Calculate the total booking value of rides completed successfully:
df[df['Booking_Status'] == 'Success']['Booking_Value'].sum()
```

```python
df[df['Booking_Status'] == 'Success'].groupby('Vehicle_Type')['Booking_Value'].mean().round(2)
```

```python
# List all incomplete rides along with the reason:
df[df['Incomplete_Rides'] == 'Yes'][['Booking_ID', 'Incomplete_Rides_Reason']]
```

```python
df['Incomplete_Rides_Reason'].unique()
```

```python
(df['Incomplete_Rides_Reason'].value_counts(normalize=True) * 100).round(2)
```
